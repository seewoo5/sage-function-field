## Method 2: Use $\mathrm{GL}_2(\mathbb{F}_q)$-action to construct bijections.

```python
# Define the right‐action via the direct sum formula
def GL2_action(A, f, n=None):
    """
    Act on f(x) in GF(p)[x] by A = [[a,b],[c,d]] via
      f|_n A := (cx + d)^n * f((ax + b)/(cx + d))
    using the equivalent sum of monomials.
    """
    R = f.parent()
    t = R.gen()
    a, b, c, d = A[0,0], A[0,1], A[1,0], A[1,1]
    if n is None:
        n = f.degree()

    num = a*t + b
    den = c*t + d

    S = sum(f.coefficient(k) * num**k * den**(n-k) for k in range(n+1))
    return S

def GL2_list(p):
    # Generate all invertible 2x2 matrices over GF(p)
    # Except for the identity matrix
    F = GF(p)
    for a in F:
        for b in F:
            for c in F:
                for d in F:
                    if a * d - b * c != 0 and (a, b, c, d) != (1, 0, 0, 1):  # Ensure the matrix is invertible
                        yield Matrix(F, [[a, b], [c, d]])

def check_invariance(m, B):
    """
    Check whether a polynomial m is invariant under the action of matrix B
    up to constant, i.e. m|_deg(m) B == c * m for some constant c in GF(p)
    """
    deg_m = m.degree()
    acted_m = GL2_action(B, m, n=deg_m)
    for c in m.parent().base_ring():
        if acted_m == c * m:
            return True, c
    return False, None

def find_order(m, gamma, delta):
    """
    Find the order of (\gamma T + delta) in (A / m)^*
    """
    R = m.parent()
    q = R.base_ring().order()
    t_ = R.gen()
    gtd = gamma * t_ + delta
    order = 0
    current = 1
    while True:
        order += 1
        current = (current * gtd) % m
        if current == 1:
            break
    return order

def exp_cong_dict(m):
    """
    Create dictionary of {e : g^e mod m} for e = 0, ..., phi(m) - 1 and its inverse
    where g is a generator of (A / m)^* computed from `unit_group_generator`
    and phi is Euler's totient function.
    """
    R = m.parent()
    q = R.base_ring().order()
    g = unit_group_generator(m)
    phi_m = euler_totient(m)
    exp_to_cong = {}
    current = R(1)
    for e in range(phi_m):
        exp_to_cong[e] = current
        current = (current * g) % m
    # cong_to_exp
    cong_to_exp = {v: k for k, v in exp_to_cong.items()}
    return exp_to_cong, cong_to_exp
```

$m(T) = T^2 + T + 1 \in \mathbb{F}_2[T]$

```python
R2.<t> = GF(2)['t']
m = t^2 + t + 1
deg_m = m.degree()

alpha = 0
beta = 1
gamma = 1
delta = 0
B = Matrix(GF(2), [[alpha, beta], [gamma, delta]])
print("Matrix B:\n", B)

assert check_invariance(m, B)[0]
N0 = find_order(m, gamma, delta)
print(f"Order of ({gamma} T + {delta}) in (F_{2}[T] / ({m}))^\\times is {N0}")

phim = euler_totient(m)
exp_to_cong, cong_to_exp = exp_cong_dict(m)

for e in range(N0):
    c_maps = []
    n = e + 3 * N0 * deg_m  # choose sufficiently large n == e mod N0
    cnt_vec = prime_count_cong_vec(n, m)
    for k in range(phim):
        c = exp_to_cong[k]
        c_act = GL2_action(B, c, n=n) % m
        k_act = cong_to_exp[c_act]
        c_maps.append((c, c_act, k, k_act))
    print(f"Ties for N == {e} mod {N0}:")
    for (c, c_act, k, k_act) in c_maps:
        assert cnt_vec[k] == cnt_vec[k_act], f"Count mismatch for c={c} and c_act={c_act}"
        print(f"  {c}  <-->  {c_act}, pi({c}, {m}, {n}) = {cnt_vec[k]} = pi({c_act}, {m}, {n})")
```
