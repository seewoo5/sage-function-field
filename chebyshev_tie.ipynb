## Method 2: Use $\mathrm{GL}_2(\mathbb{F}_q)$-action to construct bijections.

```python
# Define the right‐action via the direct sum formula
def GL2_action(A, f):
    """
    Act on f(x) in GF(p)[x] by A = [[a,b],[c,d]] via
      (cx + d)^n * f((ax + b)/(cx + d))
    using the equivalent sum of monomials.
    """
    R = f.parent()
    t = R.gen()
    a, b, c, d = A[0,0], A[0,1], A[1,0], A[1,1]
    n = f.degree()

    num = a*t + b
    den = c*t + d

    S = sum(f.coefficient(k) * num**k * den**(n-k) for k in range(n+1))
    return S

def GL2_list(p):
    # Generate all invertible 2x2 matrices over GF(p)
    # Except for the identity matrix
    F = GF(p)
    for a in F:
        for b in F:
            for c in F:
                for d in F:
                    if a * d - b * c != 0 and (a, b, c, d) != (1, 0, 0, 1):  # Ensure the matrix is invertible
                        yield Matrix(F, [[a, b], [c, d]])
```

```python
N = 3
only_irreducible = True  # Set to False to include reducible polynomials
for p in [5]:
    F = GF(p)
    R.<x> = PolynomialRing(F)
    
    print(f"Over F{p}")
    for n in range(1, N + 1):
        for f in R.monics(n):
            if only_irreducible and not f.is_irreducible():
                continue
            for g in GL2_list(p):
                result = GL2_action(g, f)
                # check if the result is a constant multiple of f
                # first extract the leading coefficient
                leading_coeff = f.leading_coefficient()
                if result == leading_coeff * f:
                    print(f"{g}")
                    print(f"f = {f}, degree = {f.degree()}")
                    print()
```
