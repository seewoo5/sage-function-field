Fibonacci polynomials $F_n(T)$ modulo 2, 3, and 5.

```python
print("Fibonacci Polynomials F_n(t) for n = 1 to 30 over F_2:")
for n in range(1, 31):
    Fn = fibo_poly(n, 2)
    m = f"F_{n}(t) = {Fn.factor()}"
    if is_perfect_power(Fn):
        m += " (perfect power)"
    if not is_perfect_power(Fn) and is_powerful(Fn):
        m += " (powerful)"
    print(m)
```
